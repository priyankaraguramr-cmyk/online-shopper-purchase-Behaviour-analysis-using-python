# Online Shoppers Purchase Intention Analysis
Domain: Sales & E-commerce

## 1. Problem Definition and Business Understanding

Online stores get a lot of visitors every day, but only a small number of them actually buy something. Every visitor who leaves without buying is a lost sale, and the money spent on bringing that visitor to the site is also wasted. So the store needs to understand what makes some visitors buy and others leave.

**Problem statement:** In this project I analyse the browsing sessions of an online store to find out which behaviours and factors are linked to a purchase, and based on that I suggest ways to improve the conversion rate.

The questions I want to answer are:
1. What percentage of sessions end in a purchase?
2. How do buyers browse differently from non-buyers (pages visited, time spent, bounce rate, exit rate, page value)?
3. Do new visitors and returning visitors behave differently?
4. Which months have the most purchases, and does a special day (like Valentine's Day) make a difference?
5. Does it matter if the visit is on a weekend, or which region, browser or traffic source the visitor comes from?
6. Which features are most related to a purchase?

This kind of analysis would be useful for the marketing and website teams of an e-commerce company, because it shows where to focus to get more sales.

## 2. Dataset Selection

For this project I chose the **Online Shoppers Purchasing Intention** dataset from the UCI Machine Learning Repository (link: https://archive.ics.uci.edu/dataset/468/online+shoppers+purchasing+intention+dataset). It was created by Sakar et al. (2018) and is based on real browsing sessions from an e-commerce website.

I chose it because it fits the project requirements well:
- It has 12,330 records, much more than the 500 required.
- It has 18 columns, which is more than the 10 required.
- It has both numerical columns (page counts, time spent, rates) and categorical columns (month, visitor type, region, traffic type, etc.), so I can do all kinds of analysis.
- It comes from a trusted source, and each row is a different user session collected over one year, so it is not biased towards one campaign or period.
- It is directly related to a real business problem (getting visitors to buy).

Columns in the dataset:
- Administrative, Informational, ProductRelated: number of pages visited in each page category
- Administrative_Duration, Informational_Duration, ProductRelated_Duration: time spent on each category (in seconds)
- BounceRates: percentage of visitors who leave after viewing only one page
- ExitRates: percentage of page views that were the last one in the session
- PageValues: average value of the pages visited before a purchase
- SpecialDay: how close the visit is to a special day (0 to 1)
- Month, OperatingSystems, Browser, Region, TrafficType: details about when and how the visitor came
- VisitorType: new, returning or other visitor
- Weekend: whether the visit was on a weekend (True/False)
- Revenue: whether the session ended in a purchase (True/False). This is the target column.
